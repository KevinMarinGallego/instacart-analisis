```python
print("Orders:", orders.shape[0])
print("Products:", products.shape[0])
print("Aisles:", aisles.shape[0])
print("Departments:", departments.shape[0])
print("Order Products:", order_products.shape[0])
```

El DataFrame con mayor cantidad de registros es order_products, porque almacena cada producto asociado a cada pedido realizado por los usuarios.

La función info() permite conocer la estructura general de un DataFrame, incluyendo el número de registros, columnas, tipos de datos y cantidad de valores nulos.

Los filtros permiten seleccionar subconjuntos específicos de información según determinadas condiciones, facilitando el análisis de los datos.

Las estadísticas descriptivas permiten resumir las características principales de una variable, mostrando medidas como promedio, valores mínimos, máximos y distribución general de los datos.



```python
order_products["reordered"].value_counts()
orders["order_hour_of_day"].mode()
(order_products["reordered"].mean()) * 100

```

Aproximadamente 59.05% de los productos registrados corresponden a compras repetidas por parte de los usuarios.
